# Practice Time

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
from nltk.tokenize import TreebankWordTokenizer
from nltk.tokenize import wordpunct_tokenize
```

```python
paragrph = """Artificial Intelligence aur Natural Language Processing aaj ke daur ki sab se important 
    technologies ban chuki hain. NLP ka use karke machines insani zubaan ko samajh sakti hain 
    aur us par kaam kar sakti hain, jaise ke text analysis, chatbots, aur language translation.
    Bohat si companies apne business ko improve karne ke liye NLP ka istemal kar rahi hain, 
    jisse customer experience behtar hota hai. Lekin! NLP ke kuch challenges bhi hain, jaise ambiguity,
    context samajhna, aur different language's ka support dena. Future mein NLP aur bhi zyada powerful ho
    jayega aur har industry mein iska use barhta chala jayega."""
```

```python
SentanceTokennize = sent_tokenize(paragrph)
SentanceTokennize
```

```python
len(SentanceTokennize)
```

```python
WordsTokennize = word_tokenize(paragrph)
WordsTokennize
```

```python
len(WordsTokennize)
```

```python
WordsPunctuationTokenize = wordpunct_tokenize(paragrph)
WordsPunctuationTokenize
```

```python
len(WordsPunctuationTokenize)
```

```python
tree_bank_word_token = TreebankWordTokenizer().tokenize(paragrph)
tree_bank_word_token
```

```python
len(tree_bank_word_token)
```
