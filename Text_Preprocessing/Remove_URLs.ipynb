# Topic03: Remove URLs

```python
# re ka matlab hai Regular Expressions.Ye Python ki built-in library hai jo pattern matching ke liye use hoti hai
import re  

text = 'Visit https://example.com or http://google.com for more info.'
print(text)
```

```python
clean_text = re.sub(r'https?://\S+|www\.\S+','',text)
print(f'Remove URLs:\nPrint only text:\n{clean_text}')
```

```python
def URLs(text):

    return re.sub(r'https?://\S+|www\.\S+','', text)

text1 = 'Visit https://example.com or http://google.com for more info.'

URLs(text1)
```
