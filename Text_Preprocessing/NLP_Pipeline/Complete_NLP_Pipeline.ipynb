# Complete NLP Pipeline

## Step:01 Required Libraries

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

# Step:02 Download only one time

```python
nltk.download('punckt')
nltk.download('stopwords')
```

```python

text = """
<html>
Visit https://example.com for more info!
This is a sample paragraph, with punctuation and stopwords.
</html>
"""
print(text)
```

# Step:03 Remove HTML Tags

```python
remove_html_tags = re.sub(r'<.*?>' , '' , text)
print(remove_html_tags)
```

# Step:04 Remove URLs

```python
remove_URLs = re.sub(r'https?://\S+|www\.\S+' , '' , remove_html_tags)
print(remove_URLs)
```

# Step:05 Remove Punctuation

```python
remove_punct = re.sub(r"[!,'.?]" , '' , remove_URLs)
print(remove_punct)
```

# Step:06 Tokenization

```python
tokenize = word_tokenize(remove_punct)
print(f'Tokens:\n{tokenize}')
```

# Step:07 Remove Stopwords

```python
remove_stopwords = set(stopwords.words('english'))
clean_tokens = [words for words in tokenize if words.lower() not in remove_stopwords]

# Final Output
print("Tokens:\nwith stopword:", tokenize)
print("\nClean Tokens:\nwithout stopwords:", clean_tokens)
print(f"\nFinal Clean Text: {" ".join(clean_tokens)}")
```
