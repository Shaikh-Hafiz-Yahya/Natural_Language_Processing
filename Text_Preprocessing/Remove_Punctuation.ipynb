# Topic: Remove Punctuation 

```python
import string , time
```

```python
punct = string.punctuation
print(f'Punctuation: {punct}') #Python Concider Punctuation
```

### This Technique Apply Very Small DataSet

```python
exclude = punct
text = 'One + Day. I, am a. successfull! person-'

def remove_punct(text):

    for char in exclude:
        
       text = text.replace(char , '')
    
    return text    

start = time.time()    
print(remove_punct(text))
time1 = time.time() - start

print(f'Time Consume is:{time1}Sec if 1 Row')
print(f'\nTime Consume is:{time1*50000}Sec if 50000 Rows')
print(f'\nTime Consume is:{time1*100000}Sec if 100000 Rows')
```
