# Topic: Remove HTML Tags

```python
import re
```

# Level 1: Basic 

### Q1: Remove HTML tags 

```python
text_with_html_tags = '<p>Hello World</p>'
print(text_with_html_tags)

remove_html_tags = re.sub(r'<.*?>','',text_with_html_tags)
print(remove_html_tags)

```

```python
text_with_html_tags14 = '<p style="color:red;">Click This Link</Link>href="https://example.com"</p>'
print(text_with_html_tags14)

remove_html_tags14 = re.sub(r'<.*?>' , '' , text_with_html_tags14)
print(remove_html_tags14)

remove_Urls = re.sub(r'https?://\S+|www\.\S+' , '' , remove_html_tags14)
print(remove_Urls)
```
