# Task 1: Perform NER on a Simple Sentence
Sentence: "Bill Gates founded Microsoft in the United States."

Steps:
Import the nltk library.

Load the English language model (en_core_web_sm).

Store the sentence in a variable.

Process the sentence using the nlp() function.

Print all detected named entities and their labels.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk
```

```python
nltk.download('maxent_ne_chunker_tab')
nltk.download('maxent_ne_chunker')   
nltk.download('words')                
nltk.download('punkt_tab')           
nltk.download('averaged_perceptron_tagger_tab') 
# sentance
sentence = "Bill Gates founded Microsoft in the United States."

# 2. Tokenization 
tokens = word_tokenize(sentence)

# 3. POS Tagging
tagged_tokens = pos_tag(tokens)

# 4. Named Entity Recognition (ne_chunk)
entities = ne_chunk(tagged_tokens)

# 5. Results print
print("\nNLTK NER Results:")

for chunk in entities:

    if hasattr(chunk, 'label'):

        entity_name = ' '.join(c[0] for c in chunk)
        entity_label = chunk.label()
        
        print(f"Entity: {entity_name:<20} | Label: {entity_label:<10}")
```
