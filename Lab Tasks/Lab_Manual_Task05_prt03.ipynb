# Task 3: Identify Specific Entity Types
#### Sentence:

"Imran Khan was born in Lahore and served as Prime Minister of Pakistan."

#### Steps
Import NLTK libraries

Download required NLTK packages

Store the sentence in a variable

Perform tokenization using word_tokenize()

Apply POS tagging using pos_tag()

Extract named entities using ne_chunk()

#### Print entities that belong to:

PERSON

LOCATION / GPE

ORGANIZATION / ORG

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk
```

```python
# sentence
sentence = "Imran Khan was born in Lahore and served as Prime Minister of Pakistan."

# Tokenization
tokens = word_tokenize(sentence)

# POS Tagging
tagged = pos_tag(tokens)

# Named Entity Recognition
entities = ne_chunk(tagged)

# Print specific entities
for entity in entities:
    if hasattr(entity, 'label'):
        entity_name = " ".join(c[0] for c in entity)
        entity_type = entity.label()

        if entity_type == 'PERSON':
            print("PERSON:", entity_name)
        elif entity_type == 'GPE':
            print("LOCATION:", entity_name)
        elif entity_type == 'ORGANIZATION':
            print("ORGANIZATION:", entity_name)
```
