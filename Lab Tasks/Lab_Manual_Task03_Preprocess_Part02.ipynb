# Task03: Text Preprocessing 02
### Write a Python program to perform the following steps:

1. Remove punctuation from the paragraph (200 words).

2. Convert into lower.

3. Turn the paragraph into tokens.

4. Perform stemming and lemmatization on a paragraph of 200 words.

```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
# Req download 
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Create Obj
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Sample Paragraph (200 words) 
paragraph = """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence. 
AI engineers are developing systems that can understand human languages perfectly. 
The students are studying various techniques like stemming and lemmatization to improve models. 
Innovation in this sector is changing the world rapidly!"""

# 1. Lowercase convert 
paragraph_lower = paragraph.lower()

# 2. Punctuation remove
cleaned_paragraph = paragraph_lower.translate(str.maketrans('', '', string.punctuation))

# 3. Convert Token
tokens = word_tokenize(cleaned_paragraph)

# 4. Perform Stemming , Lemmatization
stemmed_words = [stemmer.stem(word) for word in tokens]
lemmatized_words = [lemmatizer.lemmatize(word) for word in tokens]

print(f"Stemmed Result:\n{stemmed_words[:15]}\n") # Before 15 Tokens

print(f"Lemmatized Result:\n{lemmatized_words[:15]}\n")
```
