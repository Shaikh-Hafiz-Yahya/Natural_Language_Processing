# Task 2: NER on User Input
### Steps:

Import nltk.

Load the English model.

Ask the user to enter a sentence using input().

Process the text using nlp().

Print each entity and its type.

### Example:
Enter a sentence: Elon Musk founded SpaceX in California.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk
```

```python
# 1.input user
user_sentence = input("Enter a sentence for NER: ") 

# 2. Tokenization
tokens = word_tokenize(user_sentence)

# 3. POS Tagging 
tagged_tokens = pos_tag(tokens)

# 4. Named Entity Recognition (ne_chunk)
entities = ne_chunk(tagged_tokens)

# 5. Results
print("\nNLTK NER Results")

for chunk in entities:

    if hasattr(chunk, 'label'):

        entity_name = ' '.join(c[0] for c in chunk)
        entity_label = chunk.label()
        
        print(f"Entity: {entity_name:<20} | Type: {entity_label:<10}")
```
