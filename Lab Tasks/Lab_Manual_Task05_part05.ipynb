```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk

paragraph = "Elon Musk is the CEO of Tesla. He lives in the United States and often visits California."

tokens = word_tokenize(paragraph)
pos_tags = pos_tag(tokens)
chunked_tree = ne_chunk(pos_tags)

print(f"{'Entity':<20} | Label")
print("-" * 35)

for subtree in chunked_tree:
    if isinstance(subtree, nltk.Tree):
        entity_name = " ".join([token[0] for token in subtree.leaves()])
        entity_label = subtree.label()
        
        print(f"{entity_name:<20} | {entity_label}")
```
