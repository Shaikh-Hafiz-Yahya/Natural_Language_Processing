# Lab 4 Task: POS Tagging
### Write a Python program to perform POS tagging.

"Sara is reading a book." Write your own sentences.

Take a sentence input from the user and display POS tags for each word.

Write a program that counts the number of nouns and verbs in a sentence. (counter function will be used) from collections import Counter.

Perform POS tagging on the sentence:

"The quick brown fox jumps over the lazy dog."

Identify: Nouns, Verbs, Adjectives.

```python
import nltk
from nltk.tokenize import word_tokenize
from collections import Counter
```

```python
# Req download 
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

# Task 1,2 User Input and POS Tagging 
user_sentence = input("Enter a sentence for POS tagging: ") 
# Example: Sara is reading a book.

tokens = word_tokenize(user_sentence)
pos_tags = nltk.pos_tag(tokens)

print("\n--- POS Tags for each word ---")
print(pos_tags)

# Task 03: Count Nouns and Verbs using Counter
# Hum tags filter karain gay: Nouns (start with NN), Verbs (start with VB)
tags_list = [tag for word, tag in pos_tags]

# Only Nouns & Verbs --> Req This logic
noun_count = sum(1 for tag in tags_list if tag.startswith('NN'))
verb_count = sum(1 for tag in tags_list if tag.startswith('VB'))

print(f"\nTotal Nouns: {noun_count}")
print(f"Total Verbs: {verb_count}")

# Task 04: Fox & Dog Sentence Analysis
fox_sentence = "The quick brown fox jumps over the lazy dog."
fox_tokens = word_tokenize(fox_sentence)
fox_tags = nltk.pos_tag(fox_tokens)

nouns = [word for word, tag in fox_tags if tag.startswith('NN')]
verbs = [word for word, tag in fox_tags if tag.startswith('VB')]
adjectives = [word for word, tag in fox_tags if tag.startswith('JJ')]

print(f"\nAnalysis for: {fox_sentence}")
print(f"Nouns: {nouns}")
print(f"Verbs: {verbs}")
print(f"Adjectives: {adjectives}")
```
