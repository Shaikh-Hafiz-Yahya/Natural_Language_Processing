# Lab Task
### Write a Python program to perform the following text preprocessing steps:

Take a sentence as input.

Convert the text into lowercase.

Remove punctuation and special characters.

Remove stop words from the sentence.

Display the cleaned text after preprocessing.

# Example Input:
Hello! I am learning Natural Language Processing.

# Expected Output:
hello learning natural language processing

```python
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
# Download Data Only First Time.
nltk.download('punkt')
nltk.download('stopwords')

def clean_text(text):
    # 1. Convert Lowercase 
    text = text.lower()
    
    # 2. Punctuation , special characters Remove
    # string.punctuation ---> symbols (! " # $ % & ' () * +)
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    # 3. Tokenization 
    words = word_tokenize(text)
    
    # 4. Stop words remove (is, am, the, etc.)
    stop_words = set(stopwords.words('english'))
    cleaned_words = [w for w in words if w not in stop_words]
    
    return " ".join(cleaned_words)

# Execute 
input_text = "Hello! I am learning Natural Language Processing."
result = clean_text(input_text)

print(f"Original: {input_text}")
print(f"Cleaned Output: {result}")
```
