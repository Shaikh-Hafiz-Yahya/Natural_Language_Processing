# Stemming 

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
words = [
    "data", "science", "machine", "learning", "artificial", "intelligence",
    "python", "model", "training", "testing", "algorithm", "analysis",
    "feature", "engineering", "prediction", "classification", "regression",
    "dataset", "neural", "network", "deep", "learning", "nlp",
    "text", "processing", "language", "understanding", "tokenization",
    "stemming", "lemmatization", "embedding", "vector", "transformer",
    "attention", "sequence", "label", "accuracy", "loss", "optimization"
]
```

```python
stemming = PorterStemmer()

for word in words:
    print(word+"----->"+stemming.stem(word))
```

```python
stemming.stem('I Love NLP')
```

```python
stemming.stem('Congratulations')
```

```python
stemming.stem('MachineLearning')
```

```python
stemming.stem('Machine Learning')
```

# Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
word_net_lemmatizer = WordNetLemmatizer()

for word in words:
    print(f'{word}----->{word_net_lemmatizer.lemmatize(word)}')
```

```python
'''
pos-Noun-n
verb-v
adjective-a
adverb-r
'''

for word in words:
    print(f'{word}----->{word_net_lemmatizer.lemmatize(word , pos='v')}')
```
